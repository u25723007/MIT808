```python
#libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re
from pathlib import Path
import plotly.express as px
```

```python
#Data loading
FILE_PATH = "/content/drive/MyDrive/Capstone/Copy of Copy of Copyright and African NLP [Database] (Responses).xlsx"
SHEET_NAME = "Updated database"

df = pd.read_excel(FILE_PATH, sheet_name=SHEET_NAME)

#check columns and shape
print("Shape:", df.shape)
print("\nColumns:")
for c in df.columns:
    print("-", c)
```

```python
#Rename coulmn for easy reference

rename_map = {
    "Provide full paper citation (e.g. Okorie, C.I., 2025. The fair dealing/fair use landscape for artificial intelligence innovation and computational research in Africa. International Review of Law, Computers & Technology, pp.1-28.)": "paper_citation",
    "Provide full abbreviated reference of the paper (e.g. Okorie 2025a)": "paper_ref",
    "Authorship geography (Where are the authors based?)": "authorship_geography",
    "Institutional stewardship (What kind of institutions led the project?)": "institutional_stewardship",
    "NLP Resource Type (What did the project create?)": "nlp_resource_type_1",
    "Languages covered (What and how many languages were covered by the project?)": "languages_covered",
    "Indicate the specific languages covered": "specific_languages",
    "Geographic focus of the project": "geographic_focus",
    "Data source type (Did the project create its own data or reuse data from existing sources?)": "data_source_type",
    "Source material type ": "source_material_type",
    "Jurisdiction of data origin": "jurisdiction_data_origin",
    "Specific countries of data origin": "specific_countries_data_origin",
    "Temporal scope [What kind of materials are they?]": "temporal_scope",
    "Copyright status of the source materials": "copyright_status",
    "Copyright work type [If any of the source material is copyright-protected, what kind?]": "copyright_work_type",
    "Licence or permission status [Was permission provided to use the copyright work?]": "licence_permission_status",
    "Legal basis for use ": "legal_basis_for_use",
    "Implicit exception reliance": "implicit_exception_reliance",
    "Cross-border assumption": "cross_border_assumption",
    "Explicit IP analysis": "explicit_ip_analysis",
    'Licensing clarity [Indicate "absent" if the source material was not copyright-protected]': "licensing_clarity",
    "Rights reservation mechanisms [For the source materials, did the copyright owners use any rights reservation mechanisms?]": "rights_reservation_source_material",
    "Specific Rights reservation mechanisms [If specific rights reservation mechanisms were used, which ones?]": "specific_rights_reservation_source_material",
    "NLP Resource Type (What did the project create?) 2": "nlp_resource_type_2",
    "Documentation Practice (How is the output documented or fixed?)": "documentation_practice",
    "Ownership or Stewardship of NLP Resource": "ownership_stewardship_resource",
    "Hosting platform for resource": "hosting_platform",
    "Funding source for project": "funding_source",
    "Commercialisation orientation": "commercialisation_orientation",
    "Language marginalisation signal": "language_marginalisation_signal",
    "Update and Governance model ": "update_governance_model",
    "Rights reservation mechanisms [Where the resource or output is published or located,  are the copyright owners using any rights reservation mechanisms?]": "rights_reservation_output",
    "Specific Rights reservation mechanisms [If specific rights reservation mechanisms were used, which ones?] 2": "specific_rights_reservation_output",
    "Notes": "notes"
}

df = df.rename(columns=rename_map)
```

```python
df.head()
```

```python
#data cleaning

#Strip spaces from string columns
for col in df.columns:
    if df[col].dtype == "object":
        df[col] = df[col].astype(str).str.strip()

#Replace common pseudo-missing values
missing_tokens = {"nan", "None", "none", "N/A", "n/a", "", " "}
for col in df.columns:
    if df[col].dtype == "object":
        df[col] = df[col].replace(list(missing_tokens), np.nan)

#Create a project identifier with abbreviated paper reference
df["project_id"] = df["paper_ref"].fillna(df["paper_citation"])

print("\nMissing values per column:")
print(df.isna().sum().sort_values(ascending=False).head(10))
```

Section 1: Data overview

```python

#number of projects
num_projects = df["project_id"].nunique()

print(f"\nTotal number of NLP projects: {num_projects}")

# total rows
print(f"Total dataset records: {len(df)}")
```

```python
#Clangueges covered
language_series = df["specific_languages"].dropna()

languages = (
    language_series
    .str.split(",")
    .explode()
    .str.strip()
)

language_counts = languages.value_counts().reset_index()
language_counts.columns = ["language", "count"]

print("Top languages:")
print(language_counts.head(10))
```

```python
#language distribution

top_languages = language_counts.sort_values("count", ascending=False).head(15)

# Create label text
top_languages["LabelText"] = top_languages["count"].apply(lambda x: f"<b>{x}</b>")

# Plot
fig = px.bar(
    top_languages,
    x="language",
    y="count",
    title="Most Represented African Languages in NLP Projects",
    labels={"language": "Language", "count": "Number of Projects"},
    text="LabelText"
)

fig.update_traces(textposition="outside")
fig.update_layout(width=850,height=500)
fig.show()
```

```python
num_languages = languages.nunique()

print(f"\nTotal unique languages represented: {num_languages}")
```

```python
#languages per project

df["num_languages_project"] = df["specific_languages"].apply(
    lambda x: len([lang.strip() for lang in str(x).split(",") if lang.strip()]) if pd.notna(x) else 0
)

#categories
def language_bin(n):
    if n >= 6:
        return "6+"
    return str(n)

df["language_count_group"] = df["num_languages_project"].apply(language_bin)

#ordering
category_order = ["0", "1", "2", "3", "4", "5", "6+"]

#Count
language_group_counts = (
    df["language_count_group"]
    .value_counts()
    .reindex(category_order, fill_value=0)
    .reset_index()
)

language_group_counts.columns = ["language_count_group", "count"]

print(language_group_counts)
```

```python
fig = px.bar(
    language_group_counts,
    x="language_count_group",
    y="count",
    title="Number of Languages per NLP Project",
    labels={
        "language_count_group": "Languages in Project",
        "count": "Number of Projects"
    },
    text="count"
)

fig.update_traces(textposition="outside")

fig.update_layout(width=800, height=500,xaxis=dict(categoryorder="array", categoryarray=category_order)
)

fig.show()
```

```python
#legal basis for use vs data source type
cross_tab = pd.crosstab(
    df["data_source_type"],
    df["legal_basis_for_use"]
).reset_index()

cross_tab_melt = cross_tab.melt(
    id_vars="data_source_type",
    var_name="legal_basis_for_use",
    value_name="count"
)

fig = px.bar(
    cross_tab_melt,
    x="data_source_type",
    y="count",
    color="legal_basis_for_use",
    barmode="group",
    title="Legal Basis for Use by Dataset Source Type",
    labels={
        "data_source_type": "Dataset Source Type",
        "legal_basis_for_use": "Legal Basis for Use",
        "count": "Number of Projects"
    },
    text="count"
)

fig.update_traces(textposition="outside")
fig.update_layout(width=600,height=550)
fig.show()
```
